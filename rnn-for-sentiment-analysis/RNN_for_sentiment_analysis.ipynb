```python
# 1.Convert to lowercase
df['review'] = df['review'].str.lower()
```

```python
df.head()
```

```python
import re
```

```python
# 2. removing the URLs
def remove_urls(text):
  text = re.sub(r'http\S+', "", text)
  return text

df['review'] = df['review'].apply(remove_urls)
```

```python
df.head()
```

```python
# remove puncutations
def remove_punctuations(text):
  text = re.sub(r"[^A-Za-z0-9\s]", "", text)
  return text

df['review'] = df['review'].apply(remove_punctuations)
```

```python
df.head()
```

```python
#remove html tags
```

```python
def remove_html(text):
  text = re.sub(r"<.*?>", "", text)
  return text

df["review"] = df["review"].apply(remove_html)
```

```python
df.head()
```

```python
#  stopwards

import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
 from nltk.tokenize import word_tokenize
 from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
  tokens = word_tokenize(text)
  stop_words = stopwords.words("english")

  for word in tokens:
    if word in stop_words:
      text = text.replace(word, "")

  return text
```

```python
df['review'] = df['review'].apply(remove_stopwords)
```

```python
df.head()
```

```python
from nltk.stem import PorterStemmer
```

```python
def stemming(text):
  ps = PorterStemmer()
  stemmed_words = []

  tokens = word_tokenize(text)
  for token in tokens:
    stemmed_token = ps.stem(token)
    stemmed_words.append(stemmed_token)

  return " ".join(stemmed_words)


df["review"] = df["review"].apply(stemming)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split (
    x, y, test_size = 0.2, random_state = 42
)
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader
```

```python
x_train
```

```python
# sparse matrix cannot use directly. convert it to numpy array --> tensors
```

```python
train_set = TensorDataset(
    torch.from_numpy(x_train.toarray()).float(),
    torch.from_numpy(y_train.values).float()
)

test_set = TensorDataset(
    torch.from_numpy(x_test.toarray()).float(),
    torch.from_numpy(y_test.values).float()
)
```

```python
train_loader = DataLoader(train_set, shuffle = True, batch_size = 64)
test_loader = DataLoader(test_set, shuffle = True, batch_size = 64)
```

```python
import torch.nn as nn
import torch.optim as optim
```

```python
class RNN(nn.Module):
  def __init__(self, input_size, hidden_state = 128, num_layers = 1):
    super().__init__()

    self.hidden_size = hidden_state
    self.num_layers = num_layers

    #RNN layer
    self.rnn = nn.RNN(input_size, self.hidden_size, num_layers, batch_first = True)

    # fully connected layer
    self.fc = nn.Linear(self.hidden_size, 1)

  def forward(self, x):
    #optional ==> shape (num of layers, batch_size, hidden size)
    h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size)

    out, _ = self.rnn(x, h0)
    # 1st value = hidden state of all the timestamps => (batch, seq_len, hidden size)
    # 2nd value - final hidden state of last timestep not required

    out = self.fc(out[:, -1, :])
    #output probability value
    return out
```

```python
input_size = x_train.shape[1]

model = RNN (input_size)

criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters())
```

```python
#Training
```

```python
epochs = 10

for epoch in range(epochs):
  model.train()

  for xb, yb in train_loader:
    optimizer.zero_grad()

    xb = xb.unsqueeze(1) #add singleton direction cause model expects input 3D

    outputs = model(xb) # (batch_size,1)

    outputs = torch.sigmoid(outputs).squeeze() # (batch_size,) => probability

    loss = criterion(outputs, yb) #compute loss
    loss.backward() #backprop
    optimizer.step() # weights update

print (f"epoch = {epoch+1}/{epochs} and loss = {loss.item()}")
```

```python
# evaluate

model.eval()

with torch.no_grad():
  correct_vals = 0
  tot_vals = 0

  for xb, yb in test_loader:
    xb = xb.unsqueeze(1)

    outputs = model(xb)
    predicted = (torch.sigmoid(outputs.squeeze())> 0.5).float()

    tot_vals += yb.size(0)
    correct_vals += (predicted == yb).sum().item()

print(f"accuracy = {correct_vals/tot_vals *100}")
```
